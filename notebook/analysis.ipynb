# CineStream — ML Analysis Notebook

This notebook walks through the full CineStream ML pipeline:
1. Data loading & cleaning
2. Genre encoding (one-hot)
3. K-Means clustering (elbow + silhouette)
4. PCA visualization
5. Supervised model training (KNN + Random Forest)
6. With vs Without cluster comparison


```python
import os, sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
%matplotlib inline
matplotlib.rcParams['figure.facecolor'] = '#1a1a1f'
matplotlib.rcParams['axes.facecolor']   = '#0f0f12'
matplotlib.rcParams['text.color']       = 'white'
matplotlib.rcParams['axes.labelcolor']  = '#aaa'
matplotlib.rcParams['xtick.color']      = '#aaa'
matplotlib.rcParams['ytick.color']      = '#aaa'

from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, silhouette_score

# Add parent dir so we can import data utilities
sys.path.insert(0, os.path.join('..', 'data'))
from download_data import download_dataset
download_dataset()

print('Setup complete')
```

## 1. Load & Clean Data

```python
DATA_DIR = os.path.join('..', 'data')
movies  = pd.read_csv(os.path.join(DATA_DIR, 'movies.csv')).dropna().drop_duplicates(subset='movieId')
ratings = pd.read_csv(os.path.join(DATA_DIR, 'ratings.csv')).dropna().drop_duplicates()

movies['year']        = movies['title'].str.extract(r'\((\d{4})\)$').astype('float')
movies['title_clean'] = movies['title'].str.replace(r'\s*\(\d{4}\)\s*$', '', regex=True)

print(f'Movies: {len(movies):,}  |  Ratings: {len(ratings):,}')
movies.head()
```

## 2. Genre One-Hot Encoding

```python
all_genres = set()
for g in movies['genres']:
    all_genres.update(g.split('|'))
all_genres.discard('(no genres listed)')
genre_list = sorted(all_genres)

for genre in genre_list:
    movies[genre] = movies['genres'].str.contains(genre, regex=False).astype(int)

X_genres = movies[genre_list].values
print(f'Genre features: {len(genre_list)}')
print(genre_list)
```

## 3. K-Means — Elbow & Silhouette

```python
k_range   = range(2, 16)
inertias  = []
sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_genres)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_genres, labels, sample_size=min(2000, len(movies))))

best_k = list(k_range)[int(np.argmax(sil_scores))]
print(f'Best k = {best_k}  |  Silhouette = {max(sil_scores):.4f}')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(list(k_range), inertias, 'o-', color='#f9e27d')
ax1.axvline(best_k, color='#f9e27d', ls='--', alpha=0.5)
ax1.set_title('Elbow Method', color='white')
ax1.set_xlabel('k'); ax1.set_ylabel('Inertia')

ax2.plot(list(k_range), sil_scores, 's-', color='#7de8f9')
ax2.axvline(best_k, color='#7de8f9', ls='--', alpha=0.5)
ax2.set_title('Silhouette Score', color='white')
ax2.set_xlabel('k'); ax2.set_ylabel('Score')
plt.tight_layout(); plt.show()
```

## 4. Fit K-Means & Name Clusters

```python
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
movies['cluster'] = kmeans.fit_predict(X_genres)

cluster_names = {}
for cid in range(best_k):
    mask   = movies['cluster'] == cid
    totals = movies.loc[mask, genre_list].sum().sort_values(ascending=False)
    top2   = list(totals.head(2).index)
    cluster_names[cid] = ' + '.join(top2)
    print(f'Cluster {cid}: {cluster_names[cid]}  ({mask.sum()} movies)')
```

## 5. PCA Visualization

```python
pca    = PCA(n_components=2, random_state=42)
coords = pca.fit_transform(X_genres)
exp    = pca.explained_variance_ratio_

fig, ax = plt.subplots(figsize=(10, 7))
colors  = plt.cm.tab20(np.linspace(0, 1, best_k))
for cid in range(best_k):
    mask = movies['cluster'].values == cid
    ax.scatter(coords[mask, 0], coords[mask, 1], c=[colors[cid]], alpha=0.5, s=15,
               label=cluster_names[cid])
ax.set_title('Movie Clusters - PCA 2D', color='white', fontsize=14)
ax.set_xlabel(f'PC1 ({exp[0]*100:.1f}%)', color='#aaa')
ax.set_ylabel(f'PC2 ({exp[1]*100:.1f}%)', color='#aaa')
ax.legend(fontsize=8, labelcolor='white', facecolor='#1a1a1f', edgecolor='#333')
plt.tight_layout(); plt.show()
```

## 6. Supervised — Feature Matrix & Model Training

```python
movie_stats = ratings.groupby('movieId')['rating'].agg(movie_avg_rating='mean', movie_rating_count='count').reset_index()
user_stats  = ratings.groupby('userId')['rating'].agg(user_avg_rating='mean').reset_index()

df = ratings.merge(movies[['movieId','cluster'] + genre_list], on='movieId')
df = df.merge(movie_stats, on='movieId').merge(user_stats, on='userId')

feature_cols    = genre_list + ['cluster','movie_avg_rating','movie_rating_count','user_avg_rating']
feature_cols_nc = genre_list + ['movie_avg_rating','movie_rating_count','user_avg_rating']

X    = df[feature_cols].values.astype(float)
X_nc = df[feature_cols_nc].values.astype(float)
y    = df['rating'].values.astype(float)

print(f'Dataset: {X.shape}  |  Target: [{y.min()},{y.max()}]')
```

```python
idx = np.arange(len(y)); np.random.seed(42); np.random.shuffle(idx)
split = int(0.8 * len(idx)); tr, te = idx[:split], idx[split:]

scaler = StandardScaler()
Xs_tr = scaler.fit_transform(X[tr]); Xs_te = scaler.transform(X[te])
Xncs_tr = StandardScaler().fit_transform(X_nc[tr])
scaler_nc2 = StandardScaler().fit(X_nc[tr])
Xncs_te = scaler_nc2.transform(X_nc[te])

results = {}

knn = KNeighborsRegressor(n_neighbors=10, weights='distance', n_jobs=-1)
knn.fit(Xs_tr, y[tr])
preds = knn.predict(Xs_te)
results['KNN + cluster'] = dict(rmse=np.sqrt(mean_squared_error(y[te],preds)), mae=mean_absolute_error(y[te],preds), r2=r2_score(y[te],preds))

knn_nc = KNeighborsRegressor(n_neighbors=10, weights='distance', n_jobs=-1)
knn_nc.fit(Xncs_tr, y[tr])
preds_nc = knn_nc.predict(Xncs_te)
results['KNN no cluster'] = dict(rmse=np.sqrt(mean_squared_error(y[te],preds_nc)), mae=mean_absolute_error(y[te],preds_nc), r2=r2_score(y[te],preds_nc))

rf = RandomForestRegressor(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
rf.fit(X[tr], y[tr])
preds_rf = rf.predict(X[te])
results['RF + cluster'] = dict(rmse=np.sqrt(mean_squared_error(y[te],preds_rf)), mae=mean_absolute_error(y[te],preds_rf), r2=r2_score(y[te],preds_rf))

rf_nc = RandomForestRegressor(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
rf_nc.fit(X_nc[tr], y[tr])
preds_rfnc = rf_nc.predict(X_nc[te])
results['RF no cluster'] = dict(rmse=np.sqrt(mean_squared_error(y[te],preds_rfnc)), mae=mean_absolute_error(y[te],preds_rfnc), r2=r2_score(y[te],preds_rfnc))

pd.DataFrame(results).T.round(4)
```

## Conclusion

The table above shows how adding the K-Means cluster ID as a feature improves both KNN and Random Forest performance (lower RMSE/MAE, higher R²). This demonstrates that **unsupervised learning (K-Means) meaningfully benefits the supervised pipeline**.
